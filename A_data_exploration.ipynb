```python
#importing the neccesary libraries
import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt





```

```python
#selecting the tickers neccesary for the project and grabbing their close prices
tickers = ['SPY', 'QQQ', 'TLT', 'GLD', 'XLE']
prices  = yf.download(
    tickers,
    start="2015-01-01",
    end="2026-01-01",
    auto_adjust=True
                      )['Close']






```

```python
#seeing the first 5 entries in the data frame
print(prices.head())


#checking the info about the dataframe
print(prices.info())
```

```python






#checking if there are any null values

print(prices.isna().sum())
```

```python





#plotting the prices

prices.plot(figsize=(12, 6))
plt.title("Asset Prices")
plt.show()
```

```python
# making Log returns

log_returns = np.log(prices/ prices.shift(1))
log_returns = log_returns.dropna()

```

```python
# saving the prices and log returns

prices.to_csv("prices.csv")
log_returns.to_csv("log_returns.csv")
```
